# Malayalam text normaliser: tests and sample run

This notebook clones the public repository, installs the package, runs the pytest suite, and prints `normalise()` output for 20 sample sentences.

Malayalam word lists have not been checked by a native speaker. See the README, section Known uncertainties.

```python
import subprocess
import sys
from pathlib import Path

REPO_URL = 'https://github.com/cksajil/malayalam-text-normaliser.git'
REPO_DIR = Path('/kaggle/working/malayalam-text-normaliser')


def run(cmd, cwd=None):
    proc = subprocess.run(cmd, cwd=cwd, capture_output=True, text=True)
    print(proc.stdout)
    print(proc.stderr[-4000:])
    return proc.returncode


if REPO_DIR.exists():
    subprocess.run(['rm', '-rf', str(REPO_DIR)], check=True)
rc = run(['git', 'clone', '--depth', '1', '--branch', 'main', REPO_URL, str(REPO_DIR)])
assert rc == 0, 'git clone failed'
print('Cloned', REPO_URL)
```

```python
rc = run([sys.executable, '-m', 'pip', 'install', '-q', str(REPO_DIR) + '[dev]'])
assert rc == 0, 'pip install failed'
print('Installed malayalam-text-normaliser with dev extras')
```

```python
test_rc = run([sys.executable, '-m', 'pytest', '-q', 'tests'], cwd=str(REPO_DIR))
print('pytest exit code:', test_rc)
```

```python
from malayalam_normaliser import normalise

SAMPLES = [
    'ഇന്ന് 15/08/2026 ആണ്',
    'ഞാൻ 2 പുസ്തകങ്ങൾ വാങ്ങി',
    'വില ₹1500 ആണ്',
    'Rs 250.75 നൽകണം',
    '45% വിദ്യാർത്ഥികൾ വിജയിച്ചു',
    'യോഗം 14:30 ന് ആരംഭിക്കും',
    'പരിപാടി 9:15am ന്',
    'ദൂരം 5 km ആണ്',
    'Dr. Rao 3 മണിക്ക് വരും',
    'No. 12 ബസ്',
    'A & B',
    'താപനില -5 ആയിരുന്നു',
    'ജനസംഖ്യ 1,00,00,000 ആണ്',
    'ഭാരം 2.5 kg',
    'പണം ₹1,50,000 ആയി',
    'തീയതി 31/02/2026 ആണ്',
    'പാഠം 0 മുതൽ 100 വരെ',
    'ചെലവ് 12.00 Rs ആയി',
    'ഫോൺ 007 ആണ്',
    'ഉച്ചയ്ക്ക് 12:00 pm, 3% പലിശ',
]

lines = ['| # | Input | Output |', '|---|---|---|']
for i, sentence in enumerate(SAMPLES, 1):
    lines.append('| %d | %s | %s |' % (i, sentence, normalise(sentence)))
table = '\n'.join(lines)
print(table)
Path('/kaggle/working/sample_table.md').write_text(table + '\n', encoding='utf-8')
```

```python
if test_rc != 0:
    raise SystemExit('pytest failed, see the log above')
print('All tests passed')
```
